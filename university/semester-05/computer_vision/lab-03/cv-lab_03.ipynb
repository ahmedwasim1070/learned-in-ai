## 2. Background Theory[cite: 1]
Geometric transformations change the spatial arrangement of pixels. A point $(x, y)$ is mapped to a new coordinate location $(x', y')$[cite: 1]. This lab covers translation, rotation, scaling, affine transformation, perspective transformation, and homography[cite: 1].

| Transformation | Main Effect | OpenCV Function[cite: 1] |
| :--- | :--- | :--- |
| **Translation** | Moves an image | `cv2.warpAffine()`[cite: 1] |
| **Rotation** | Changes orientation | `cv2.getRotationMatrix2D()` + `cv2.warpAffine()`[cite: 1] |
| **Scaling** | Changes image size | `cv2.resize()`[cite: 1] |
| **Affine** | Combines translation/rotation/scaling/shearing | `cv2.getAffineTransform()` + `cv2.warpAffine()`[cite: 1] |
| **Perspective** | Models viewpoint/perspective change | `cv2.getPerspectiveTransform()` + `cv2.warpPerspective()`[cite: 1] |
| **Homography** | Maps corresponding points between planes | $3 \times 3$ projective matrix $H$[cite: 1] |

## 6. Part B — Translation[cite: 1]
Translation shifts an image horizontally by $t_x$ and vertically by $t_y$[cite: 1]. The affine transformation matrix is:
$$
M = \begin{bmatrix} 1 & 0 & t_x \\ 0 & 1 & t_y \end{bmatrix}
$$
- Run with $t_x = 100, t_y = 50$[cite: 1].
- Run with $t_x = -100, t_y = 80$[cite: 1].
- Observe direction and cropping behavior[cite: 1].

````python
# Translation 1: tx = 100, ty = 50
tx1, ty1 = 100, 50
M1 = np.float32([
    [1, 0, tx1],
    [0, 1, ty1]
])
translated1 = cv2.warpAffine(image, M1, (width, height))

# Translation 2: tx = -100, ty = 80
tx2, ty2 = -100, 80
M2 = np.float32([
    [1, 0, tx2],
    [0, 1, ty2]
])
translated2 = cv2.warpAffine(image, M2, (width, height))

# Display outputs
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(translated1, cv2.COLOR_BGR2RGB))
plt.title("Translated: tx=100, ty=50")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(translated2, cv2.COLOR_BGR2RGB))
plt.title("Translated: tx=-100, ty=80")
plt.axis("off")
plt.show()
````

### Part B: Observation[cite: 1]
- **Direction of Shift:** Positive $t_x$ shifts the image rightwards, whereas negative $t_x$ shifts it leftwards. Positive $t_y$ shifts the image downwards, and negative $t_y$ shifts it upwards (aligned with OpenCV's coordinate origin at top-left $(0, 0)$).
- **Cropping & Padding:** Since the output canvas size is fixed to `(width, height)`, shifted content that exceeds canvas boundaries is permanently clipped. Newly uncovered regions inside the canvas boundary are filled with black pixels (zeros).

## 9. Part E — Affine Transformation[cite: 1]
An affine transformation preserves points, straight lines, and parallelism[cite: 1]. It has 6 degrees of freedom and requires exactly **3 non-collinear point correspondences** to solve the $2 \times 3$ transformation matrix[cite: 1]:
$$
\begin{bmatrix} x' \\ y' \end{bmatrix} = \begin{bmatrix} a_{00} & a_{01} \\ a_{10} & a_{11} \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} + \begin{bmatrix} b_0 \\ b_1 \end{bmatrix}
$$

````python
pts1 = np.float32([
    [50, 50],
    [200, 50],
    [50, 200]
])

pts2 = np.float32([
    [80, 70],
    [220, 60],
    [60, 220]
])

M_affine = cv2.getAffineTransform(pts1, pts2)
print("Affine Transformation Matrix (2x3):")
print(M_affine)

affine = cv2.warpAffine(image, M_affine, (width, height))

plt.figure(figsize=(6, 5))
plt.imshow(cv2.cvtColor(affine, cv2.COLOR_BGR2RGB))
plt.title("Affine Transformation")
plt.axis("off")
plt.show()
````

### Part E: Observation & Explanation[cite: 1]
- **Inspection of Matrix:** The resulting $2 \times 3$ matrix $M = [A \mid B]$ contains the linear transformation coefficients in the left $2 \times 2$ block (accounting for rotation, non-uniform scaling, and shear), while the right column $[b_0, b_1]^T$ specifies translational displacement.
- **Effect of Changing Destination Points:** Changing destination points $pts_2$ directly alters the affine mapping parameters. Moving one point while holding others fixed shears the plane, stretching parallel lines diagonally while preserving their parallelism.

## 10. Part F — Perspective Transformation[cite: 1]
Perspective transformation models projective foreshortening where parallel lines can converge[cite: 1]. It has 8 degrees of freedom and requires **4 corresponding coplanar points**[cite: 1].

````python
src = np.float32([
    [100, 100],
    [500, 80],
    [520, 400],
    [80, 420]
])

dst = np.float32([
    [0, 0],
    [400, 0],
    [400, 500],
    [0, 500]
])

H_persp = cv2.getPerspectiveTransform(src, dst)
print("Homography / Perspective Matrix H (3x3):")
print(H_persp)

corrected = cv2.warpPerspective(image, H_persp, (400, 500))

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(corrected, cv2.COLOR_BGR2RGB))
plt.title("Perspective Corrected Image")
plt.axis("off")
plt.show()
````

### Part F: Correspondence Roles[cite: 1]
- **What the 4 Source Points Represent:** The 4 pixel coordinates of an arbitrary distorted quadrilateral or slanted plane in the input image[cite: 1].
- **What the 4 Destination Points Represent:** The target rectangular viewing boundary (top-left, top-right, bottom-right, bottom-left) mapped to canonical orthographic coordinates $(0, 0) \rightarrow (W, H)$[cite: 1].
- **Modifying a Source Point:** Moving any single source coordinate skews the projective mapping, distorting the local magnification and angle of the transformed plane.

## 11. Part G — Homography and Corresponding Points[cite: 1]
A planar homography $H$ maps homogeneous points between two projective planes[cite: 1]:
$$
\begin{bmatrix} x' \\ y' \\ w' \end{bmatrix} = \begin{bmatrix} h_{11} & h_{12} & h_{13} \\ h_{21} & h_{22} & h_{23} \\ h_{31} & h_{32} & h_{33} \end{bmatrix} \begin{bmatrix} x \\ y \\ 1 \end{bmatrix}[cite: 1]
$$
The inhomogeneous Cartesian coordinates are recovered via perspective division[cite: 1]:
$$
X' = \frac{x'}{w'}, \quad Y' = \frac{y'}{w'}[cite: 1]
$$

| Point | Source Point $(x, y)$ | Destination Point $(x', y')$[cite: 1] |
| :--- | :--- | :--- |
| **P1** | (100, 100) | (0, 0)[cite: 1] |
| **P2** | (500, 80) | (400, 0)[cite: 1] |
| **P3** | (520, 400) | (400, 500)[cite: 1] |
| **P4** | (80, 420) | (0, 500)[cite: 1] |

### Part G: Analytical Questions & Answers[cite: 1]

1. **Why are four point correspondences used for a general homography?**[cite: 1]
   - **Answer:** A $3 \times 3$ homography matrix has 9 entries, but because homogeneous coordinates are defined up to an arbitrary non-zero scale factor, it has **8 independent degrees of freedom** (DoF). Each point correspondence provides 2 independent linear constraints ($x'$ and $y'$ equations). Therefore, $\frac{8}{2} = 4$ non-collinear point pairs are mathematically necessary and sufficient to uniquely solve for $H$[cite: 1].

2. **What is the role of the third row of the homography matrix?**[cite: 1]
   - **Answer:** The third row $[h_{31}, h_{32}, h_{33}]$ governs **projective foreshortening and depth scaling**. While an affine matrix fixes its third row to $[0, 0, 1]$, non-zero values for $h_{31}$ and $h_{32}$ cause $w'$ to vary as a function of spatial position $(x, y)$, causing parallel lines to converge and modeling perspective tilt.

3. **Why are transformed coordinates normalized using $w'$?**[cite: 1]
   - **Answer:** In projective geometry, homogeneous coordinates $[x', y', w']^T$ represent an equivalence ray in projective space $\mathbb{P}^2$. To map this back into the standard 2D physical Cartesian screen space $\mathbb{R}^2$, we must project onto the canonical plane $w'=1$ via perspective division: $X' = x'/w'$ and $Y' = y'/w'$[cite: 1].

4. **What is the difference between estimating $H$ and applying $H$?**[cite: 1]
   - **Answer:** **Estimating $H$** is an inverse problem: calculating the numerical parameters of the $3 \times 3$ matrix given a set of known point pairs using Direct Linear Transformation (DLT). **Applying $H$** is a forward/warping operation: using a known matrix $H$ with backward mapping and interpolation (`cv2.warpPerspective`) to compute output pixel intensities for the whole image[cite: 1].

## 12. Part H — Main Practical Task: Perspective Correction[cite: 1]
Select an image containing a planar rectangular object (document, card, sign)[cite: 1]. Identify its four corners, compute the homography, and apply perspective correction to generate a fronto-parallel rectified view[cite: 1].

````python
# Load image containing a planar object (using sample.jpg as demo)
doc_image = cv2.imread("sample.jpg")
if doc_image is None:
    raise FileNotFoundError("Please provide a valid image for Part H.")

# Coordinates of the 4 corners of the planar document in sample.jpg
# Order: Top-Left, Top-Right, Bottom-Right, Bottom-Left
src_points = np.float32([
    [120, 80],   # Top-Left
    [680, 80],   # Top-Right
    [680, 520],  # Bottom-Right
    [120, 520]   # Bottom-Left
])

# Target dimensions for the rectified orthographic document
out_w, out_h = 500, 400
dst_points = np.float32([
    [0, 0],
    [out_w, 0],
    [out_w, out_h],
    [0, out_h]
])

# 1. Estimate homography matrix
H_doc = cv2.getPerspectiveTransform(src_points, dst_points)
print("Estimated Homography Matrix H:")
print(H_doc)

# 2. Warp perspective to rectify
doc_corrected = cv2.warpPerspective(doc_image, H_doc, (out_w, out_h))

# 3. Save output
cv2.imwrite("perspective_corrected.jpg", doc_corrected)

# 4. Display comparison
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
img_with_pts = doc_image.copy()
for pt in src_points.astype(int):
    cv2.circle(img_with_pts, tuple(pt), 10, (0, 0, 255), -1)
plt.imshow(cv2.cvtColor(img_with_pts, cv2.COLOR_BGR2RGB))
plt.title("Original with Corner Markers")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(doc_corrected, cv2.COLOR_BGR2RGB))
plt.title("Perspective Corrected (Fronto-Parallel View)")
plt.axis("off")
plt.show()
````

## 16. Lab Questions[cite: 1]

1. **What is the main difference between a geometric transformation and an intensity-based image filter?**[cite: 1]
   - **Answer:** A geometric transformation maps spatial pixel coordinate locations $(x, y) \rightarrow (x', y')$, changing pixel arrangement without modifying intrinsic gray/color values[cite: 1]. An intensity filter modifies pixel values based on intensity functions or local neighborhoods while preserving the spatial coordinate grid.

2. **What is the effect of translation parameters $t_x$ and $t_y$?**[cite: 1]
   - **Answer:** $t_x$ shifts the image horizontally (positive shifts right, negative shifts left)[cite: 1]. $t_y$ shifts the image vertically (positive shifts downward, negative shifts upward in standard image matrices)[cite: 1].

3. **What is the difference between uniform and non-uniform scaling?**[cite: 1]
   - **Answer:** Uniform scaling uses equal scale factors ($f_x = f_y$), preserving aspect ratio, angles, and relative proportions[cite: 1]. Non-uniform scaling uses unequal factors ($f_x \neq f_y$), which stretches or compresses the image along one axis and distorts shapes[cite: 1].

4. **What type of transformation is represented by a $2 \times 3$ affine matrix?**[cite: 1]
   - **Answer:** It represents a 6-DoF linear mapping that includes translation, rotation, scaling, and shear[cite: 1]. It preserves straight lines, collinearity, and parallelism[cite: 1].

5. **How many point correspondences are minimally required for a general homography?**[cite: 1]
   - **Answer:** Exactly **4 coplanar point correspondences** (no 3 collinear) because homography has 8 degrees of freedom and each point provides 2 constraints[cite: 1].

6. **Why does a projective transformation use homogeneous coordinates?**[cite: 1]
   - **Answer:** Homogeneous coordinates allow non-linear projective operations (including translation and perspective foreshortening) to be represented and computed as linear matrix multiplications[cite: 1].

7. **What is the purpose of $w'$ in the projective transformation?**[cite: 1]
   - **Answer:** $w'$ represents the projective scaling factor associated with depth[cite: 1]. Normalizing coordinates by dividing by $w'$ ($X'=x'/w', Y'=y'/w'$) projects the homogeneous ray back to physical Cartesian 2D coordinates[cite: 1].

8. **Why can parallel lines appear to converge after a perspective transformation?**[cite: 1]
   - **Answer:** Because projective transformations model pinhole camera geometry where real-world parallel lines meet at a vanishing point on the horizon line due to depth foreshortening[cite: 1].

9. **What is the difference between an affine transformation and a projective transformation?**[cite: 1]
   - **Answer:** Affine transformations (6 DoF) keep the line at infinity fixed and **preserve parallelism**[cite: 1]. Projective transformations (8 DoF) can map the line at infinity to a finite line, **failing to preserve parallelism** and modeling convergence[cite: 1].

10. **What practical problem does perspective correction solve?**[cite: 1]
    - **Answer:** It rectifies images taken from slanted or arbitrary viewing angles into top-down, fronto-parallel scans (crucial for document OCR, license plate recognition, and sports field analysis)[cite: 1].

## 17. Challenge Task: Interactive Point Selection[cite: 1]
The script below uses Matplotlib's event handler to let the user click 4 corner points directly on the image, then computes the homography and rectifies the view[cite: 1].

````python
class InteractivePerspective:
    def __init__(self, img_path):
        self.img = cv2.imread(img_path)
        if self.img is None:
            raise FileNotFoundError(f"Cannot load {img_path}")
        self.points = []

    def manual_demo(self, clicked_pts):
        """Executes rectification with selected 4 coordinates."""
        src = np.float32(clicked_pts)
        w, h = 450, 600
        dst = np.float32([[0, 0], [w, 0], [w, h], [0, h]])
        
        H = cv2.getPerspectiveTransform(src, dst)
        warped = cv2.warpPerspective(self.img, H, (w, h))
        
        plt.figure(figsize=(10, 5))
        plt.subplot(1, 2, 1)
        display_img = self.img.copy()
        for i, pt in enumerate(src.astype(int)):
            cv2.circle(display_img, tuple(pt), 8, (0, 255, 0), -1)
            cv2.putText(display_img, f"P{i+1}", tuple(pt + 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
        plt.imshow(cv2.cvtColor(display_img, cv2.COLOR_BGR2RGB))
        plt.title("Selected 4 Corners")
        plt.axis("off")
        
        plt.subplot(1, 2, 2)
        plt.imshow(cv2.cvtColor(warped, cv2.COLOR_BGR2RGB))
        plt.title("Interactive Rectified View")
        plt.axis("off")
        plt.show()
        return H

# Run challenge demo with 4 ordered corner clicks (TL, TR, BR, BL)
runner = InteractivePerspective("sample.jpg")
clicked = [[120, 80], [680, 80], [680, 520], [120, 520]]
H_interactive = runner.manual_demo(clicked)
````
